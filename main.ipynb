```python
# 12 Read all the files from the bronze csv's and dump metadata of file like filename , whene the file was loaded , no of rows in file and colnames
# 13. Create a file-level validation report: file exists, schema columns present, and row count.


import os,datetime,shutil

files_to_be_removed=["silver","gold","silver_rejected","file_validation_report.csv"]

for file in files_to_be_removed:
    if os.path.exists(file):
    
        if file.endswith(".csv"):
            os.remove(file)
        else:
            shutil.rmtree(file)

path=f"bronze"

list_of_dict=[]

files=os.listdir(path)
for index,file in enumerate(files):
       try:
           if file.endswith(".csv"):
               list_of_dict.append({})
               list_of_dict[-1]["file_name"]=file
               df=pd.read_csv(f'{path}/{file}')
               list_of_dict[-1]["row_count"]=df.shape[0]
               list_of_dict[-1]["load_time"]=datetime.datetime.now().timestamp()
               list_of_dict[-1]["columns"]=",".join(list(df.columns))
       except Exception as e:
           print(e)

df=pd.DataFrame(list_of_dict)
df.to_csv("file_validation_report.csv")
```

```python
# 44. Build `gold_monthly_sales`: month, order_count, gross_revenue, discount_amount, net_revenue.
merged['month']=pd.to_datetime(merged['order_date']).dt.month
merged
```

```python
df=merged.groupby('month').agg(
    monthly_transactions=("transaction_id","count"),
    total_gross_revenue=("gross_amt","sum"),
    total_disc_amount=("disc_amt","sum"),
    total_net_revenue=("total_amt","sum")
)
df.to_csv("gold/gold_monthly_sales.csv",index=False)
```

```python
# 45. Build `gold_category_sales`: category, order_count, net_revenue, revenue_rank.
df=merged.groupby('category').agg(
    monthly_transactions=("transaction_id","count"),
    total_gross_revenue=("gross_amt","sum"),
    total_disc_amount=("disc_amt","sum"),
    total_net_revenue=("total_amt","sum")
)
df['net_rev_rank']=df['total_net_revenue'].rank(method="dense",ascending=False)
#df.to_csv("gold/gold_monthly_sales.csv",index=False)
df
```

```python
#46. Build `gold_customer_360`: customer, city, segment, completed_order_count, lifetime_revenue, last_order_date.

df=merged[ merged['order_status'] == "Completed"  ].groupby(['customer_name','city','customer_segment']).agg(
    completed_order_count = ("transaction_id","count")
)

df= df.merge( merged.groupby('customer_name')['order_date'].max() , on ="customer_name",how="left")
df=df.merge( merged.groupby('customer_name')['total_amt'].sum() , on ="customer_name",how="left" )
df.rename({"order_date":"last_order_date","total_amt":"lifetime_spend"},axis=1,inplace=True)
df.to_csv("gold/gold_customer_360.csv")
```

```python
# 47. Build `gold_order_status_summary`: status, order_count, percentage_of_orders.
df=merged.groupby('order_status').agg(
    order_count=("transaction_id","count"),
)
order_count= int (merged['transaction_id'].count())
df['order_percentage']=df['order_count']/order_count *100
df.to_csv("gold/gold_order_status_summary.csv")
```

```python
# 48. Build `gold_daily_sales`: order_date, completed_orders, net_revenue, running_revenue.

df=merged[merged['order_status']=="Completed"].groupby(["order_date"]).agg(
    completed_order_count=("transaction_id","count")
)
df=df.merge( merged.groupby(["order_date"]).agg( net_revenue=("total_amt","sum")  ) , on=["order_date"], how="outer" )
df.sort_values("order_date",ascending=True,inplace=True)
df['running_net_revenue']=df['net_revenue'].cumsum()
df['completed_order_count']=df['completed_order_count'].fillna(0)
df.to_csv("gold/gold_daily_sales.csv")
df
```
